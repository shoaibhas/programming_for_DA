```python
import pandas as pd
import matplotlib.pyplot as plt
url = "https://raw.githubusercontent.com/plotly/datasets/master/gapminderDataFiveYear.csv"
df = pd.read_csv(url)
for country in ["Germany","India","Pakistan"]:
    rowa = df[df["country"] == country]
    plt.plot(rowa["year"], rowa["lifeExp"],label = country)

plt.legend()
plt.show()


latest = df[df["year"] == 2007]
medians = latest.groupby("continent")["lifeExp"].median().sort_values()
medians.plot(kind = "barh", colormap = "Set2")
plt.title("Median country life expectancy")
plt.xlabel("Life expectancy (years)")
plt.show()


plt.scatter(latest["gdpPercap"],latest["lifeExp"],alpha = 0.6)
plt.xscale("log")
plt.xlabel("GDP per person")
plt.ylabel("life expectancy (years)")
plt.title("GDP per person and life expectancy, 2007")
plt.show()
```
